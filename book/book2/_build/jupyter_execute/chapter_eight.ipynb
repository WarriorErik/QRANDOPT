## Sidebar Controls Explained

- **🔢 Number of raw bits to generate**: Determines how many random bits (0s and 1s) the dashboard generates for analysis. More bits give a more accurate statistical picture, but take longer to compute.
- **📦 Batch size for Aer fallback**: When true quantum randomness isn't available, the dashboard simulates quantum bits using Qiskit Aer in batches of this size.
- **📏 Minimum length for SP 800-22 tests**: The NIST randomness tests require a certain minimum bit length; this slider sets that threshold.
- **🔊 Bit-flip noise probability**: Simulates noise by flipping each bit with this probability, helping you see how noise affects randomness metrics.


## Example: Effect of Number of Raw Bits
Let's see how increasing the number of raw bits smooths out the bias estimate.

```python
import numpy as np
import matplotlib.pyplot as plt

def estimate_bias(num_bits, trials=50):
    biases = []
    for _ in range(trials):
        bits = np.random.choice([0, 1], size=num_bits)
        bias = abs(bits.mean() - 0.5)
        biases.append(bias)
    return biases

# Estimate bias for different sample sizes
sizes = [100, 1000, 10000]
plt.figure()
for n in sizes:
    biases = estimate_bias(n)
    plt.plot(biases, label=f"{n} bits")
plt.title("Bias Estimates vs. Number of Bits")
plt.xlabel("Trial")
plt.ylabel("Absolute Bias")
plt.legend()
plt.show()
```

## What is Q-Learning?

Q-Learning is a type of **Reinforcement Learning** where an agent learns to make decisions by interacting with an environment and receiving rewards. Over time, it builds a **Q-table** that estimates the value of taking a certain action in a given state.

Key terms:
- **State**: A description of the current situation (e.g., the current bias and entropy).
- **Action**: A choice the agent can make (e.g., pick a specific extractor).
- **Reward**: Feedback from the environment (higher if the extracted bits are more random).
- **Q-table**: A lookup table where each entry Q[state, action] estimates the expected future reward of taking that action in that state.
