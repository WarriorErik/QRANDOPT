Welcome to the very first chapter of the QRANDOPT Jupyter Book! In this chapter, we'll explain in simple terms what the QRANDOPT dashboard does and why randomness matters.

## Why Randomness Matters

Randomness plays a crucial role in many fields:
- **Cryptography:** Secure communication depends on unpredictable keys.
- **Simulations:** Accurate modeling of real-world processes often requires random sampling.
- **Scientific experiments:** Random trials help remove bias.

The QRANDOPT dashboard helps you generate, analyze, and improve random bitstreams.

## Generating a Raw Bitstream

The dashboard can generate random bits either from a public randomness beacon or a quantum simulator. Here’s a simple Python example using NumPy to generate bits:


```python
import numpy as np

#generate 20 random bits with equal probability
bits = np.random.choice([0,1], size=20)
bits.tolist()
```

This list of 0s and 1s represents our **raw bitstream**. The dashboard uses more advanced sources (like quantum hardware), but the concept is the same.

## Measuring Bias in a Bitstream

A perfectly random bitstream has an equal number of 0s and 1s. Bias measures how far we are from that ideal:

```python
bias = abs(np.mean(bits) - 0.5)
bias
```

If `bias` is `0.0`, the stream is perfectly balanced. Larger values indicate more imbalance.
